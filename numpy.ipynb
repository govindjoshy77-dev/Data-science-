# arange()

it is similar to python's range(), but it returns a Numpy array

np.arange(start,stop,step)
* start ----> Where to start
* stop  ----> Where to Stop(not included)
* step  ----> Difference between values


```
arr=np.arange(1,10,2)
print(arr)
```

```
arr=np.arange(10)
```

```
arr
```

```
arr=np.arange(1,100)
```

```
print(arr)
```

# linspace()

It creates numbers that are equally spaced between a start and an end value.

np.linspace(start,stop,num).
* start ---> starting value.
* stop  ---> ending value.
* num   ---> how many values you want.

```
arr=np.linspace(0,10,5)
```

```
print(arr)
```

```
arr=np.linspace(0,10,5,endpoint=True) ### by default endpoint True
print(arr)
```

```
arr=np.linspace(0,10,5,endpoint=False) ### remove the last element (10)
print(arr)
```
